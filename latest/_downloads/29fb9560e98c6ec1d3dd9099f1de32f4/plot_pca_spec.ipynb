# PCA analysis example

In this example, we perform the PCA dimensionality reduction of a
spectra dataset


Import the spectrochempy API package


```python
import spectrochempy as scp
```

# Load and inspect the dataset


```python
dataset = scp.read_omnic("irdata/nh4y-activation.spg")[::5]
dataset
```

```python
_ = dataset.plot()
```

# PCA on the full spectral range

Create a PCA object and fit the dataset so that the explained variance
is greater or equal to 99.9%


```python
pca = scp.PCA(n_components=0.999)
_ = pca.fit(dataset)
```

The number of retained components is given by the `n_components`
attribute. Because only 11 spectra were kept (`[::5]`), the rank of the
(centered) dataset is at most 10, so at most 10 principal components
have nonzero variance. The threshold `n_components=0.999` retains the
number shown below:


```python
pca.n_components
```

Transform the dataset to a lower dimensionality using all the fitted
components


```python
scores = pca.transform()
scores
```

# Display the results graphically

First we can set some preferences for the plot


```python
prefs = scp.preferences
prefs.lines.markersize = 7

# ScreePlot
_ = pca.plot_scree()
```

Score Plot first we can set some preferences for the plot


```python
prefs.lines.markersize = 10

_ = pca.plot_score()
```

Score Plot for 3 PC\'s in 3D


```python
_ = pca.plot_score(components=(1, 2, 3))
```

Displays 4 loadings


```python
_ = pca.loadings[:4].plot(legend=True)
```

# Mask the saturated region and refit

Here we do a masking of the saturated region between 882 and 1280 cm\^-1


```python
dataset[
    :, 882.0:1280.0
] = scp.MASKED  # remember: use float numbers for slicing (not integer)
_ = dataset.plot()
```

Apply the PCA model


```python
pca = scp.PCA(n_components=0.999)
_ = pca.fit(dataset)
pca.n_components
```

After masking the saturated region, fewer components are needed to reach
the same 99.9% variance threshold, as shown below. The masked interval
(882 to 1280 cm\^-1) is dominated by a strongly saturated (truncated)
detector response: masking it removes a large, non-informative source of
variance, so the number of retained components decreases accordingly.
The saturated band is not chemically meaningful, so it must be excluded
before the analysis.


```python
_ = pca.plot_scree()
```

Note that the number of retained PCA components is set by the variance
threshold. It is neither the effective rank of the dataset nor the
number of independent chemical species: `PCA(n_components=0.999)` simply
keeps the smallest number of principal components whose cumulative
variance reaches 99.9%.


Displays the loadings


```python
_ = pca.loadings.plot(legend=True)
```

Let\'s plot the scores


```python
scores = pca.transform()
_ = pca.plot_score((1, 2))
```

# Label the score plot with custom labels

Our dataset has already two columns of labels for the spectra but there
are a bit too long for display on plots.


```python
scores.y.labels
```

So we define some short labels for each component, and add them as a
third column:


```python
labels = [lab[:6] for lab in dataset.y.labels[:, 1]]
scores.y.labels = labels  # Note this does not replace previous labels,
# but adds a column.
```

# Display the labeled score plot

Labels are now placed automatically using the `adjustText` library,
which provides collision avoidance between labels and markers. Install
`adjustText` with `pip install adjustText` for best results; without it,
labels are still shown with a simple offset from markers.


```python
_ = pca.plot_score(scores=scores, show_labels=True, labels_column=2)
```
